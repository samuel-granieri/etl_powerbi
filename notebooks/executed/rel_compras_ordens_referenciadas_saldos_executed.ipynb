## Transformação de dados - rel_compras_ordens_referenciadas_saldos

#### 1.Carregar tabela do bronze

```python
import sys
sys.path.append("/app")


from utils import create_spark_session, load_config, save_table
from pyspark.sql import functions as F
from delta.tables import DeltaTable

tabela_nome = "rel_compras_ordens_referenciadas_saldos"

spark = create_spark_session("rel_compras_ordens_referenciadas_saldos")
config = load_config()

# ler bronze
df = spark.read.format("delta").load(
    f"data/bronze/{tabela_nome}"
)
```

#### 2.Executar transformação

```python

# transformação
df_silver = (
    df
    .select("ordem_compra_id", "produto_id", "id_item", "empresa_id", "saldo", "id")
)
```

#### 3.Armazenar dados na camada silver

```python
# salvar silver
path = f"data/silver/{tabela_nome}"

if DeltaTable.isDeltaTable(spark, path):

    delta_table = DeltaTable.forPath(spark, path)

    (
        delta_table.alias("t")
        .merge(
            df_silver.alias("s"),
            """
            t.id = s.id
            AND t.produto_id = s.produto_id
            AND t.empresa_id = s.empresa_id
            AND t.ordem_compra_id = s.ordem_compra_id
            AND t.id_item = s.id_item
            """
        )
        .whenMatchedUpdateAll()
        .whenNotMatchedInsertAll()
        .execute()
    )

else:

    df_silver.write.format("delta").save(path)
```
